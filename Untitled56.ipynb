```python
print("Membaca data dari database SQLite...")
db_path = '/content/drive/MyDrive/superstore (1).sqlite'
conn = sqlite3.connect(db_path)

# Query bersih tanpa JOIN ke tabel returns
query_clf = """
    SELECT
        oi.sales, oi.discount, oi.quantity, oi.shipping_cost,
        o.order_key, o.ship_mode, o.order_priority,
        c.segment,
        l.market, l.region,
        p.category, p.sub_category
    FROM order_items oi
    JOIN orders o ON oi.order_key = o.order_key
    JOIN dim_customers c ON o.customer_id = c.customer_id
    JOIN dim_locations l ON o.location_id = l.location_id
    JOIN dim_products p ON oi.product_id = p.product_id
"""
df_clf = pd.read_sql_query(query_clf, conn)
conn.close()
```

PEMBUATAN TARGET & PREPROCESSING

```python
df_clf['is_urgent'] = df_clf['order_priority'].apply(lambda x: 1 if x in ['Critical', 'High'] else 0)

print("Distribusi Kelas (0 = Normal, 1 = Urgent):")
print(df_clf['is_urgent'].value_counts())

# Buang kolom yang tidak boleh ditebak oleh model atau hanya berupa ID
X_clf = df_clf.drop(columns=['is_urgent', 'order_priority', 'order_key'])
y_clf = df_clf['is_urgent']
```

FEATURE ENGINEERING

```python
X_clf['shipping_cost_ratio'] = X_clf['shipping_cost'] / X_clf['sales']
X_clf['discount_impact'] = X_clf['discount'] * X_clf['sales']
X_clf['total_cost_spent'] = X_clf['sales'] + X_clf['shipping_cost']
```

SPLIT & ENCODING

```python
import numpy as np

# Handle infinite and NaN values that may have resulted from feature engineering (e.g., division by zero).
# It's best to apply this fix to the original X_clf before encoding and splitting.
for col in X_clf.select_dtypes(include=np.number).columns:
    # Replace NaN values with 0 (or another appropriate imputation strategy like median/mean)
    if X_clf[col].isnull().any():
        X_clf[col] = X_clf[col].fillna(0)
    # Replace infinite values with 0
    if (X_clf[col] == np.inf).any() or (X_clf[col] == -np.inf).any():
        X_clf[col] = X_clf[col].replace([np.inf, -np.inf], 0)

cat_cols_clf = X_clf.select_dtypes(include=['object']).columns.tolist()
X_encoded_clf = pd.get_dummies(X_clf, columns=cat_cols_clf, drop_first=True)

X_train_clf, X_test_clf, y_train_clf, y_test_clf = train_test_split(
    X_encoded_clf, y_clf, test_size=0.2, random_state=42, stratify=y_clf
)

# Define numerical columns to be scaled. We take them from X_clf to ensure we only scale
# the original numerical and engineered features, excluding the one-hot encoded binary features.
num_cols_clf = X_clf.select_dtypes(exclude=['object']).columns.tolist()

scaler_clf = StandardScaler()

X_train_scaled_clf = X_train_clf.copy()
X_test_scaled_clf = X_test_clf.copy()

X_train_scaled_clf[num_cols_clf] = scaler_clf.fit_transform(X_train_clf[num_cols_clf])
X_test_scaled_clf[num_cols_clf] = scaler_clf.transform(X_test_clf[num_cols_clf])
```

TRAINING MODEL XGBOOST

```python
print("\nMelatih model XGBoost...")
xgb_classifier_urgent = XGBClassifier(
    random_state=42,
    eval_metric='logloss',
    n_estimators=200,
    learning_rate=0.1,
    max_depth=5
)

xgb_classifier_urgent.fit(X_train_scaled_clf, y_train_clf)
```

```python
y_pred_clf = xgb_classifier_urgent.predict(X_test_scaled_clf)
y_pred_proba = xgb_classifier_urgent.predict_proba(X_test_scaled_clf)[:, 1]

print("\n--- HASIL EVALUASI KLASIFIKASI PRIORITAS (URGENT VS NORMAL) ---")
print(f"Akurasi Keseluruhan: {accuracy_score(y_test_clf, y_pred_clf) * 100:.2f}%")
print(f"Skor ROC-AUC       : {roc_auc_score(y_test_clf, y_pred_proba):.4f}\n")

print("Classification Report:")
print(classification_report(y_test_clf, y_pred_clf, target_names=['Normal (0)', 'Urgent (1)']))

# Confusion Matrix
cm = confusion_matrix(y_test_clf, y_pred_clf)
plt.figure(figsize=(6, 4))
sns.heatmap(cm, annot=True, fmt='d', cmap='Oranges',
            xticklabels=['Prediksi Normal', 'Prediksi Urgent'],
            yticklabels=['Asli Normal', 'Asli Urgent'])
plt.title('Confusion Matrix - Deteksi Prioritas Pesanan')
plt.ylabel('Aktual')
plt.xlabel('Prediksi')
plt.show()
```

```python
import pandas as pd

# ==========================================
# 1. FORM INPUT UI COLAB (Klasifikasi Prioritas)
# (Ubah input di panel sebelah kanan cell ini!)
# ==========================================
# @title 📦 Detektor Prioritas Pesanan Gudang

Sales_USD = 200000 # @param {type:"number"}
Diskon_Persen = 77 # @param {type:"slider", min:0, max:100, step:1}
Biaya_Kirim_USD = 350 # @param {type:"number"}
Kuantitas = 7 # @param {type:"integer"}

Kategori = 'Furniture' # @param ["Furniture", "Office Supplies", "Technology"]
Sub_Kategori = 'Appliances' # @param ["Accessories", "Appliances", "Art", "Binders", "Bookcases", "Chairs", "Copiers", "Envelopes", "Fasteners", "Furnishings", "Labels", "Machines", "Paper", "Phones", "Storage", "Supplies", "Tables"]
Segment = 'Corporate' # @param ["Consumer", "Corporate", "Home Office"]
Region = 'Southeast Asia' # @param ["Central", "North", "South", "East", "West", "Southeast Asia", "Oceania", "EMEA", "Africa", "Canada", "Caribbean", "Central Asia", "North Asia"]
Market = 'APAC' # @param ["APAC", "EU", "US", "LATAM", "EMEA", "Africa", "Canada"]
Ship_Mode = 'First Class' # @param ["Standard Class", "Second Class", "First Class", "Same Day"]

# Masukkan ke dictionary (Order Priority TIDAK ADA karena itu yang ditebak)
user_input_urgent = {
    'sales': [Sales_USD],
    'discount': [Diskon_Persen / 100.0],
    'quantity': [Kuantitas],
    'shipping_cost': [Biaya_Kirim_USD],
    'ship_mode': [Ship_Mode],
    'segment': [Segment],
    'market': [Market],
    'region': [Region],
    'category': [Kategori],
    'sub_category': [Sub_Kategori]
}

# ==========================================
# 2. PROSES PREDIKSI
# ==========================================
df_input_urgent = pd.DataFrame(user_input_urgent)

# Feature Engineering
df_input_urgent['shipping_cost_ratio'] = df_input_urgent['shipping_cost'] / df_input_urgent['sales']
df_input_urgent['discount_impact'] = df_input_urgent['discount'] * df_input_urgent['sales']
df_input_urgent['total_cost_spent'] = df_input_urgent['sales'] + df_input_urgent['shipping_cost']

# Encoding & Penyesuaian Kolom
df_input_encoded = pd.get_dummies(df_input_urgent)
training_columns_urgent = X_train_scaled_clf.columns
df_input_encoded = df_input_encoded.reindex(columns=training_columns_urgent, fill_value=0)

# Scaling menggunakan scaler_clf
df_input_scaled = df_input_encoded.copy()
df_input_scaled[num_cols_clf] = scaler_clf.transform(df_input_encoded[num_cols_clf])

# Prediksi Kelas (0 atau 1)
prediksi_prioritas = xgb_classifier_urgent.predict(df_input_scaled)[0]

# ==========================================
# 3. TAMPILAN OUTPUT
# ==========================================
print("==========================================")
print(f"🎯 OUTPUT MODEL: {prediksi_prioritas}")
print("==========================================")

if prediksi_prioritas == 1:
    print("🚨 KESIMPULAN: URGENT (1) - Pesanan ini berisiko tinggi / butuh penanganan VIP!")
else:
    print("✅ KESIMPULAN: NORMAL (0) - Pesanan ini bisa diproses masuk antrean reguler gudang.")
```
